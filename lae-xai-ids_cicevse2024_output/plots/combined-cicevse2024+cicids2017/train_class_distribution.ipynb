```python
import matplotlib.pyplot as plt
import matplotlib as mpl
import numpy as np

mpl.rcParams.update({
    'font.family': 'serif', 'font.serif': ['Times New Roman', 'DejaVu Serif'],
    'font.size': 15, 'axes.labelsize': 17, 'axes.titlesize': 17,
    'xtick.labelsize': 14, 'ytick.labelsize': 14, 'legend.fontsize': 13,
    'axes.linewidth': 1.5, 'xtick.major.width': 1.5, 'ytick.major.width': 1.5,
    'xtick.direction': 'in', 'ytick.direction': 'in',
    'xtick.top': True, 'ytick.right': True,
    'figure.dpi': 150, 'savefig.dpi': 300, 'mathtext.fontset': 'stix',
})

evse_classes = ['Aggressive\nscan', 'ICMP flood /\nfragmentation', 'Identity rotation\n& rotation flood',
                'Normal', 'OS finger-\nprinting', 'Port\nscanning', 'Service\ndetection',
                'Slow request\nstarvation', 'Stealth SYN\nscanning', 'TCP\nfloods',
                'UDP\nflood', 'Vulnerability\nscan']
evse_train_full    = [82018, 72, 209748, 65, 112218, 341327, 234285, 3361, 424757, 576835, 25980, 185094]
evse_train_sampled = [1640, 72, 4194, 65, 2244, 6826, 4685, 3361, 8495, 11536, 519, 3701]
evse_test_eval      = [20501, 8, 52436, 17, 28053, 85313, 58561, 840, 106185, 144209, 6495, 46270]
evse_test_full      = [20501, 8, 52436, 17, 28053, 85313, 58561, 840, 106185, 144209, 6495, 46270]

ids_classes = ['BENIGN', 'Bot', 'Brute\nForce', 'DoS/\nDDoS', 'Infiltration', 'PortScan', 'Web\nAttack']
ids_train_full    = [1818477, 1573, 11068, 304559, 29, 127144, 1744]
ids_train_sampled = [36369, 1573, 221, 6091, 29, 2542, 1744]
ids_test_eval      = [2000, 246, 1991, 2000, 7, 486, 403]
ids_test_full      = [378637, 246, 1991, 65207, 7, 486, 403]

colors = ['#1f77b4', '#d62728', '#2ca02c', '#9467bd']
markers = ['o', 's', '^', 'D']
labels = ['Train (full)', 'Train (sampled)', 'Test (eval)', 'Test (full, disjoint)']

fig, axes = plt.subplots(1, 2, figsize=(20, 8.5))
datasets = [
    (evse_classes, [evse_train_full, evse_train_sampled, evse_test_eval, evse_test_full]),
    (ids_classes, [ids_train_full, ids_train_sampled, ids_test_eval, ids_test_full]),
]

handles = None
for ax, (classes, series_list) in zip(axes, datasets):
    x = np.arange(len(classes))
    for series, color, marker, lab in zip(series_list, colors, markers, labels):
        ax.plot(x, series, marker=marker, markersize=8, linewidth=2.2,
                 color=color, label=lab, markeredgecolor='black', markeredgewidth=0.6)
    ax.set_yscale('log')
    ax.set_xticks(x); ax.set_xticklabels(classes, rotation=45, ha='right', fontsize=12)
    ax.set_xlim(-0.4, len(classes) - 0.6)
    ax.set_ylabel('Number of Samples')
    ax.tick_params(which='both', direction='in', top=True, right=True)
    for spine in ax.spines.values(): spine.set_linewidth(1.5)
    ax.grid(True, which='major', linestyle='--', linewidth=0.5, alpha=0.4)
    if handles is None:
        handles, legend_labels = ax.get_legend_handles_labels()

fig.legend(handles, legend_labels, loc='upper center', bbox_to_anchor=(0.5, 1.0),
           ncol=4, frameon=False, handletextpad=0.6, columnspacing=1.8)

for ax, lab in zip(axes, ['(a) CICEVSE2024', '(b) CICIDS2017']):
    ax.text(0.5, -0.20, lab, transform=ax.transAxes, ha='center', va='top',
            fontsize=16, fontweight='bold')

plt.subplots_adjust(top=0.90, bottom=0.30, left=0.055, right=0.99, wspace=0.15)
plt.savefig('class_distribution_line.png', pad_inches=0.15)
plt.show()
```
